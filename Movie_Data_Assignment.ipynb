### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
drama = data[data['Genre'].str.contains('Drama', na=False)]['Gross'].dropna()
action = data[data['Genre'].str.contains('Action', na=False)]['Gross'].dropna()

plt.boxplot([drama, action], tick_labels=['Drama', 'Action'])

plt.title('Gross Revenue: Drama vs Action')
plt.ylabel('Gross Revenue')
plt.show()
```
